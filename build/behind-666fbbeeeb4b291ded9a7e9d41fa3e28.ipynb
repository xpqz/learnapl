The _Behind_ operator (sometimes also referred to as _Reverse Compose_, or the Σ-combinator, if you are boring at parties) was introduced in the v20 release of Dyalog APL. Like _Over_ we met earlier, it's one of those patterns that seems pointless until the penny drops, and suddenly you see applications of it everywhere. Conor Hoekstra, \@code_report, dubbed it (somewhat tongue-in-cheek, perhaps) the [most significant addition](https://www.youtube.com/watch?v=recwy_MVgho) to Dyalog APL since the introduction of trains in 2014.

_Behind_ is easy enough to model in APL yourself:

```apl
Behind ← {⍺←⍵ ⋄ (⍺⍺ ⍺)⍵⍵ ⍵} ⍝ ⍛
```

So what does it do? In homage to Conor, here's his example:

```apl
' ' ∘≠⍛/ 'mr owl ate my metal worm' ⍝ Remove spaces
```

That's saying "compress by the is-not-space binary mask", which we could have spelled out as:

```apl
{⍵/⍨' '≠⍵} 'mr owl ate my metal worm' ⍝ Remove spaces
```

One way to think of _Behind_ is as a predicate-applicator. With compress, it becomes a filter (`⍛/`), as just saw. If you "say" it, and omit the _Behind_, it becomes "not-equal-to-space compress". Here's the "first equals" predicate application to find the places where the first element of a vector occurs. Say it, spell it:

```apl
⊃⍛= 2 7 1 8 2 8 1 8 2 8 4 5 9 0 4 5 ⍝ first equals
```

We can 'stack' Behinds, too, combining the predicate application and filter aspects. From Adam's video above, return all integers from a vector:

```apl
⌊⍛= 2.1 ¯5 99 3.14 4 1.3 ⍝ floor equals
```

```apl
⌊⍛=⍛/ 2.1 ¯5 99 3.14 4 1.3 ⍝ floor equals compress
```
