The _Behind_ operator (sometimes also referred to as _Reverse Compose_, or the Σ-combinator) was introduced in the v20 release of Dyalog APL. Like _Over_ we met earlier, it's one of those patterns that seems pointless until the penny drops, and suddenly you see applications of it everywhere. Conor Hoekstra, \@code_report, dubbed it (somewhat tongue-in-cheek, perhaps) the [most significant addition](https://www.youtube.com/watch?v=recwy_MVgho) to Dyalog APL since the introduction of trains in 2014.

_Behind_ is easy enough to model in APL yourself:

```apl
Behind ← {⍺←⍵ ⋄ (⍺⍺ ⍺)⍵⍵ ⍵} ⍝ ⍛
```

So what does it do? In homage to Conor, here's his example:

```apl
' ' ∘≠⍛/ 'mr owl ate my metal worm' ⍝ Remove spaces
```

That's saying "compress by the is-not-space binary mask", which we could have spelled out as:

```apl
{⍵/⍨' '≠⍵} 'mr owl ate my metal worm' ⍝ Remove spaces
```

One way to think of _Behind_ is as a predicate-applicator, or filter. If you "say" it, and omit the _Behind_, it becomes "not-equal-to-space-compress".
