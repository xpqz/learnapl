# Ride

> Keats and Yeats are on your side / While Wilde is on mine.
\--_Morrissey_ [@morrissey1986cemetry]

`Ride` started life as a [Microsoft Silverlight](https://en.wikipedia.org/wiki/Microsoft_Silverlight) application, and was intended to become the one <s>Ring</s> IDE to rule them all, in terms of Dyalog's supported platforms. It didn't turn out that way. It was since rewritten in JavaScript, as an [Electron](https://www.electronjs.org/) app and is an [open source project](https://github.com/dyalog/ride). `Ride`'s using the [Monaco editor component](https://microsoft.github.io/monaco-editor/) that also powers VS Code.

`Ride` has two distinct roles, sometimes conflicting. On the one hand, it's used to attach to remote, running instances of Dyalog to facilitate debugging of running applications. On the other hand, it's the default IDE that non-Windows users use to develop Dyalog APL locally on their computers. As the "non-Windows users" segment of Dyalog's user base has always been comparatively small, `Ride` occasionally feels not entirely comfortable on macOS or linux. Especially on macOS, it doesn't always respect the established OS conventions, particularly when it comes to how it handles multiple windows. There is little we, as users, can do about that, except let Dyalog know that this makes the interface sluggish and the "feel" wrong. 

The reason it's like this is that the "classic" model for the Windows IDE is that of Windows 95 -- each application opens separate windows for everything, and this doesn't really fit with the application model of Electron -- or indeed macOS, for that matter. If you're a classic-mode Dyalog user on Windows and you need `Ride` to debug a remote session, the model feels comfortable, and this is the root of the uncanny valley that makes `Ride` feel a bit awkward. Multiple sessions in `Ride` are completely separate application processes, and on macOS, you'll then see many `Ride` icons in your dock, instead of a single icon which gathers its session windows like you'd probably expect. 

Other issues -- that we _can_ partly address -- are for example the out of box configuration settings that are both hard to discover and -- quite frankly -- set to ridiculous defaults. The other aspect is the philosophy: `Ride` _looks_ much like any other editor/IDE you will be familiar with, but is better seen as a "dumb" display tool. It delegates _everything_ to the interpreter, including the state of windows and tabs. It reflects the workspace model of Dyalog APL. It has no concept of "files", or "save/open/close" in the ways you expect. Without Link, which we met earlier, using `Ride` would be a pretty tiresome experience to me. Once we understand the workspace model, and how Link maps this to the file system, we can begin to make the most of `Ride`. And don't get me wrong -- these issues really are paper cuts, not deal breakers.

We've talked about workspaces and Link already. Let's dwell a bit on the session interface -- in `Ride` specifically, but coming back to the "dumb display tool" idea, is actually mostly the interpreter itself dictating the behaviour. The session is really an active canvas, and does not at all behave like the shell interface it looks like. If you expect to be able to use the arrow keys to recall your session history, you'll be surprised to see that instead it moves the cursor up and down your previous output. If you press return, it tries to re-execute whatever line you happen to be on -- including lines that cannot be executed. This plays badly with boxed output, and the beginner will frequently end up in a state like this:

![ride1](./IMG/ride-1.png)

A long, boxed output, and you tried moving up a few lines and pressed return, expecting shell behaviour, but got a canvas. Perhaps you then tried to clear the screen from all the junk output with a <kbd>ctrl</kbd>-<kbd>l</kbd>, but that had no effect. The ability to re-evaluate lines in the output flow is something APLers love and see as a feature. It takes a while to adapt to, if you used to normal editors, shells and repls. You _can_, of course, also go up and down your session history, just not with the arrows, by default. Instead, you use <kbd>ctrl</kbd>-<kbd>shift</kbd>-<kbd>backspace</kbd> for up, and <kbd>ctrl</kbd>-<kbd>shift</kbd>-<kbd>enter</kbd> for down. That's worth committing to memory immediately. 

Let's look at some quality of life configuration changes. Click on the little keyboard symbol in the top right to open `Ride`'s preferences settings, and select the `Shortcuts` tab. You should see something like this:

:::{image} ./IMG/ride-prefs.png
:alt: The Shortcuts tab of Ride's preferences window
:width: 70%
:align: center
:::

This tab lets you view, modify and create new keyboard bindings. First, in the search box, type `FX`, Dyalog-speak for what the rest of the world would think of as `Save`. Click the little `+` symbol, and enter the key-combo that corresponds to your operating system's standard for `Save` -- on macOS, <kbd>⌘s</kbd>. Click `APPLY`. Next, search for `SI` (Strong Interrupt), and set that to <kbd>ctrl</kbd>-<kbd>c</kbd>. 

It's worth botanising a bit in this window -- a lot of functionality is only available on shortcuts that have no default key setting, and some have cryptic names to boot.
