> Namespace is a (class 9) object in Dyalog APL. Namespaces are analogous to nested workspaces.

except that Dyalog's namespaces don't obey the time-honoured name scoping rules you expect. Sadly.

## Namespaces as dicts

A namespace in Dyalog APL can be thought of as a dictionary mapping keys to values. Values are accessed by the dot notation that should feel familiar for those of you who know an object-oriented language like JavaScript or Python. 

There are many ways to construct namespaces in Dyalog APL. The simplest is perhaps to use APLAN,

````apl
⊢ ns ← (
  author: 'Gödel, Kurt'
  title: 'On Formally Undecidable Propositions of Principia Mathematica and Related Systems I'
  booktitle: 'Kurt Gödel: Collected Works, Volume I'
  publisher: 'Oxford University Press'
  year: 1931
  note: 'trans. Jean van Heijenoort, publ. 1986'
)
````

Namespaces don't display, and APLers see this as a feature (and if you think of the other use of namespaces in APL -- actual namespaces -- you can probably see why). The namespace notation in APLAN is very close to that of many other languages, e.g. Python:

```python
~/work/notebooks/learnapl/contents [20] $ python
Python 3.11.3 (main, Apr 28 2025, 10:54:41) [Clang 16.0.0 (clang-1600.0.26.6)] on darwin
Type "help", "copyright", "credits" or "license" for more information.
>>> {
...   'author': 'Gödel, Kurt',
...   'title': 'On Formally Undecidable Propositions of Principia Mathematica and Related Systems I',
...   'booktitle': 'Kurt Gödel: Collected Works, Volume I',
...   'publisher': 'Oxford University Press',
...   'year': 1931,
...   'note': 'trans. Jean van Heijenoort, publ. 1986'
... }
{'author': 'Gödel, Kurt', 'title': 'On Formally Undecidable Propositions of Principia Mathematica and Related Systems I', 'booktitle': 'Kurt Gödel: Collected Works, Volume I', 'publisher': 'Oxford University Press', 'year': 1931, 'note': 'trans. Jean van Heijenoort, publ. 1986'}
>>>
```

Python does have a nice display form for dicts -- as they _are_ dicts, not namespaces masquerading as dicts.

We can view our namespace by converting it to JSON:

````apl
NSShow ← 1∘(⎕JSON⍠'Compact' 0)
NSShow ns
````

### Members are names, not strings

One important distinction between our namespaces and the keys in an actual dict is that Dyalog APL's namespace keys are APL _names_, not strings. This comes back to the central duality. A "key" in a namespace is really an APL variable, and so it can sometimes help to think of a namespace as an object instance. This is perhaps clearer if we look at a different way of constructing namespaces, using [⎕NS](https://dyalog.github.io/documentation/21.0/language-reference-guide/system-functions/ns/). 

````apl
ns ← ⎕NS⍬ ⍝ Construct empty anonymous namespace
ns.author ← 'Gödel, Kurt'
ns.title ← 'On Formally Undecidable Propositions of Principia Mathematica and Related Systems I'
ns.booktitle ← 'Kurt Gödel: Collected Works, Volume I'
ns.publisher ← 'Oxford University Press'
ns.year ← 1931
ns.note ← 'trans. Jean van Heijenoort, publ. 1986'
````

````apl
NSShow ns
````

If we think about the namespace interpretation of the namespace (a phrase that sounds more vacuous every time I type it), it's not a big leap to see that we can make use of APL's normal assignment rules here:

````apl
values ← (
    'Gödel, Kurt'
    'On Formally Undecidable Propositions of Principia Mathematica and Related Systems I'
    'Kurt Gödel: Collected Works, Volume I'
    'Oxford University Press'
    1931
    'trans. Jean van Heijenoort, publ. 1986'
)

ns ← ⎕NS⍬
ns.(author title booktitle publisher year note) ← values

NSShow ns
````

In Python, dict keys can be anything that's hashable, but in Dyalog APL, namespace members must have valid APL names -- they are identifiers. Dyalog jumps through hoops in order to smooth over this wrinkle. We'll see more about this when we introduce `⎕JSON` properly, but JSON, of course, assumes that keys are values:

````apl
NSShow ns ← 0⎕JSON'{"First name": "Kurt", "Last name": "Gödel"}'
````

So far, so 100% expected. But `⎕JSON` does something under the covers, which we can uncover by actually requesting the keys:

````apl
]map ns
````

`⎕JSON` seamlessly mangles and un-mangles keys to be valid namespace members. Thus, to access a single key, we need to do

````apl
ns.⍙First⍙32⍙name
````

### Getting and setting values with `⎕VGET` and `⎕VSET`

Even accepting the name vs value quirk for keys, using namespaces as dicts in Dyalog used to be both clunky and slow, but in the v20 release of Dyalog APL, several core changes made it much more ergonomic and performant. We've already seen how APLAN makes it very natural to compose dict literals. Let's first show some of the problematic aspects. If we have a vector of keys and a vector of values, how can we associate those in a namespace? Prior to v20, we'd have to construct assignments as strings that were then evaluated inside the namespace:

````apl
ns ← ⎕NS⍬

keys ← (
    'author'
    'title'
    'booktitle'
    'publisher'
    'year'
    'note'
)

values ← (
    'Gödel, Kurt'
    'On Formally Undecidable Propositions of Principia Mathematica and Related Systems I'
    'Kurt Gödel: Collected Works, Volume I'
    'Oxford University Press'
    1931
    'trans. Jean van Heijenoort, publ. 1986'
)

⍝ Dyalog APL pre-v20
ns {⍺.{⍎⍺,'←⍵'}¨/⍵} keys values 
⍝   │  ││    │ └───────────────── For every key-value pair,
⍝   │  │└────┴─────────────────── the string '{key}←{val}' is
⍝   │  └─────────────────────────    ...evaluated
⍝   └────────────────────────────    ...inside the namespace

NSShow ns
````

Works, sure. But if you think that is ugly, unsafe and inefficient, that's because it _is_. Imagine if a key contained `⎕OFF⋄`.

Fortunately, there is now a nicer, safer and faster way, using the system function [⎕VSET](https://dyalog.github.io/documentation/21.0/language-reference-guide/system-functions/vset/):

````apl
ns ← ⎕NS⍬
ns ⎕VSET (↑keys) values  ⍝ Dyalog v20+
NSShow ns
````

Why the _Mix_ you say? Well, that's so that `⎕VSET` can also take a zipped list of key-value pairs:

````apl
ns ← ⎕NS⍬

⍝ A subset of the data...
ns ⎕VSET ('author' 'Gödel, Kurt' ⋄ 'title' 'On Formally Undecidable Propositions of Principia Mathematica and Related Systems I')
NSShow ns
````

The _Mix_ in the first instance is necessary in order for `⎕VSET` to be able to distinguish between the two different forms of its right argument.

Getting values as strings suffered the analogous problem to the setting which we outlined above. `⎕VSET` has a sibling function, [⎕VGET](https://dyalog.github.io/documentation/21.0/language-reference-guide/system-functions/vget/), that solves that for us:

````apl
keys ← 'author' 'title'

ns ⎕VGET keys
````

`⎕VGET` has a very nice feature built in, similar to how Python's `.get()` method works on dicts -- we can specify default values to be returned if a key isn't found:

````apl
ns ⎕VGET ('author' '-- no author given --' ⋄ 'title' '-- no title given --' ⋄ 'year' '-- no year given --')
````

For a single key with default, we need to enclose the right argument:

````apl
ns ⎕VGET ⊂'year' '-- no year given --'
````

The combination of APLAN and `⎕VSET|VGET` has some powerful, but not immediately obvious consequences. An empty parenthesis pair is now APLAN for "new namespace". That means we can say

````apl
keys ← 'artist' 'country' 'language' 'genre'
vals ← 'Drake' 'Canada' 'English' 'Hip-Hop'

ns ← ()⎕VSET(↑keys)vals

NSShow ns
````

### Arrays of namespaces

Namespace references -- like any other APL type -- can be arranged in arrays, and this has some neat consequences. For example, dot-access applies to an array of any rank:

````apl
jazz ← (
  (title: 'Kind of Blue'              ⋄ artist: 'Miles Davis'     ⋄ year: 1959)
  (title: 'The Shape of Jazz to Come' ⋄ artist: 'Ornette Coleman' ⋄ year: 1959)
  (title: 'A Love Supreme'            ⋄ artist: 'John Coltrane'   ⋄ year: 1965)
  (title: 'Mingus Ah Um'              ⋄ artist: 'Charles Mingus'  ⋄ year: 1959)
  (title: 'Head Hunters'              ⋄ artist: 'Herbie Hancock'  ⋄ year: 1973)
)

jazz.artist
````

````apl
(1959=jazz.year)/jazz.title
````

This also extents to `⎕VGET` and `⎕VSET`:

````apl
(1959=jazz⎕VGET'year')/jazz⎕VGET'title'
````

We can construct some SQL-like helper operators here:

````apl
_Where_ ← {⍵ ⍵⍵ ⍺⍺ ⎕VGET ⍺}
_Select_ ← {⍵/⍺⍺ ⎕VGET ⍵⍵}
````

````apl
(jazz _Select_ 'title') 'year' (jazz _Where_=) 1959
````

We can add a column to all member namespaces at once, as `⎕VSET` happily takes array left arguments:

````apl
jazz ⎕VSET ⊂'genre' 'jazz'
````

````apl
⍪NSShow¨jazz
````

### Reference semantics

As we alluded to above, namespaces are _references_ in Dyalog APL. This has some interesting consequences, such as mutation via `⍺` (or `⍵`) in a dfn:

````apl
kob ← (title: 'Kind of Blue' ⋄ artist: 'Miles Davis' ⋄ year: 1959)

SetGenre ← { ⍺ ⎕VSET ⊂'genre' ⍵ }

kob SetGenre 'Jazz'

NSShow kob
````

References also lead to aliasing,

````apl
a ← (foo: 'bar')
b ← a
b ⎕VSET ⊂'fee' 'baz' ⍝ Mutate `a` via its `b` alias

NSShow a
````

which is a behaviour we don't see with values:

````apl
a ← 1 2 3 4
b ← a         ⍝ Copy
b ,← 99

a
````

It's easy to slip up here. Take a look at this:

````apl
(b a) ← ⎕NS⍬
a.foo ← 'bar'
b.foo ← 'baz'

(a b).foo
````

What happened there was that we created a SINGLE namespace reference, which then got scalar-extended to the two variables `a` and `b` -- with `b` now aliasing `a`.

Dyalog APL has several reference types, also including objects (closely related to namespaces). Mutable references has a tendency to add "spice and excitement" to your code, so use wisely. 

If you want value semantics for namespaces, you need to "deep clone" them. `⎕NS` can take a namespace right argument, in which case it will return a deep clone:

````apl
a ← (foo: 'bar')
b ← ⎕NS a            ⍝ `b` is a deep clone of `a`
b ⎕VSET ⊂'fee' 'baz' ⍝ Mutate `b` only

⍪NSShow¨a b
````

We can use the cloning facility of `⎕NS` to merge namespaces, which gives us a neat way of implementing progressive defaults. `⎕NS` can take any number of namespaces to the right, and return a new namespace which is the union (with last-wins replacement):

````apl
defaults ← (role: 'Drummer' ⋄ name: 'None given')
person ← ⎕NS defaults (name: 'Baard Kolstad')
NSShow person
````

## Namespaces as <s>namespaces</s> structured containers for code

The other aspect of the unhelpful duality of Dyalog APL namespaces is as a code container construct -- an actual namespace. In its simplest form, it requires but a small mental leap: values in a namespace can be functions as well as arrays:

````apl
ns ← ()

ns.Sum ← +/

ns.Sum ⍳100
````

In this form, if you squint a bit, a Dyalog APL namespace start to resemble a JavaScript object. We can mock up the beginnings of a maths utility library:

````apl
Math ← ()
Math.Abs ← |
Math.Cos ← 2∘○
Math.Pi ← ○1

Math.Cos Math.Pi
````

Sadly, we can't (yet) write this as APLAN:

````apl
Math ← (    ⍝ SYNTAX ERROR: Function and operator elements are not permitted (v21)
  Abs: |
  Cos: 2∘○
  Pi: ○1
)
````

but rumours are that this extension is on the roadmap for a future Dyalog release.

Dyalog relies extensively on this type of namespaces in pretty much all APL code it ships. Namespaces are one of the core mechanisms for modularity in APL code, but if you know more mainstream programming languages, some aspects will not work as you expect -- but that's kind of true for APL in general, isn't it. 

### Namespaces and variable scope

Namespaces and variable scope varies greatly between programming languages. Let's illustrate a few scope gotchas. Here's a contrived example implementing a counter:

````apl
Counter ← (count: 0)
Counter.Increment ← {
    count +← 1             ⍝ I can see names defined in my own namespace
    count
}

count ← 100                ⍝ Outer variable, lexically scoped

Demo ← {
    count ← 5              ⍝ Shadows the outer `count`
    _ ← Counter.Increment⍬
    ⎕ ← count              ⍝ 5 -- inner scope wins
    ⎕ ← Counter.count      ⍝ 1 -- reached by named access
    ⍬
}

Demo ⍬
count                      ⍝ 100 -- never touched
````

That was hopefully what you expected to see. What if we add another layer of namespace nesting? 

````apl
Counter ← (count: 0)
Counter.(inner ← ())
Counter.inner.Increment ← {
    count +← 1                ⍝ VALUE ERROR: Undefined name: count
    count
}

Counter.inner.Increment ⍬
````

If that was what you expected to see, well done -- another deep insight. In other words, in Dyalog APL, namespaces are NOT lexical scopes. In order for the `inner` namespace to see names defined in its enclosing namespace, we must reference those explicitly:

````apl
Counter ← (count: 0)
Counter.(inner ← ())
Counter.inner.Increment ← {
    ##.count +← 1
    ##.count
}

Counter.inner.Increment ⍬
````

The `##.` thing means "...in the enclosing namespace". Note that the following does NOT work:

````apl
Counter ← (count2: 0)
Counter.inner ← ()
Counter.inner.Increment ← {
    ##.count2 +← 1                
    ##.count2
}

Counter.inner.Increment ⍬
````

Why? The key is the lack of parentheses in `Counter.inner ← ()`. This does _not_ create a new namespace in the `Counter` namespace under the name `inner`, it creates a namespace (reference) in the root namespace, and references that as the name `inner` in `Counter`. 

A single `#` refers to the global root namespace, and `⎕THIS` returns a reference to the current namespace. With that knowledge at hand, try to explain the following behaviour:

````apl
(foo←{⎕THIS})⍬

(⎕SE.foo←{⎕THIS})⍬

⎕SE.foo⍬
````

[⎕SE](https://dyalog.github.io/documentation/21.0/language-reference-guide/system-functions/se/) is the SEssion namespace, a place where various system-y things get loaded on startup. The last two invocations call the exact same instance of the `foo` function; they ought to return the same thing, but they don't. Bug or feature?

## Named namespaces

So far, we've created anonymous namespace references on the fly, primarily via the `()` APLAN convenience. There are several other ways to create namespaces. We can create _named_ namespaces instead of anonymous references:

````apl
⎕ ← 'bar'⎕NS⍬
````

````apl
ns ← ()
⎕NC 'ns'
⎕NC '#.bar'
````

## Namespaces as nested workspaces

We started this chapter with an admittedly poor attempt at a joke, dissing Dyalog's claim that

> Namespace is a (class 9) object in Dyalog APL. Namespaces are analogous to nested workspaces. 

is somewhat opaque. However, it's worth exploring. We've glossed over the "workspace" concept at best so far, partly because it's a model that doesn't really appear much in today's mainstream development environments. We can think of a Dyalog APL workspace as a binary snapshot containing your defined functions and variables etc. This is a concept that's been part of APL implementations since its inception. It has some seductive qualities: as a binary snapshot, it's easily frozen in time and distributable. It has little start-up cost. It Just Works Everywhere. Workspaces used to be completely flat, with zero structure, and at some stage, someone decided that it would be helpful to have the ability to impose some structure. Namespaces are one solution to this -- and understanding

Binary workspaces, of course, have drawbacks, too -- which is why we rarely see them outside APL (and [Smalltalk](https://en.wikipedia.org/wiki/Smalltalk)) today. In 2026, programmers expect to be able to store their source code as textfiles in a versioning system, like [git](https://git-scm.com/). That's potentially awkward in a workspace-oriented paradigm. Dyalog's [Link](https://dyalog.github.io/link) is a response to this: it maps a directory of source files into a Dyalog APL workspace as a set of nested namespaces. We'll discuss Link in a later chapter. 

Without a great leap of imagination, we can think of a workspace with nested namespaces as conceptually similar to a hierarchical filesystem with `#` representing the root. We can use the system function [⎕NL](https://dyalog.github.io/documentation/21.0/language-reference-guide/system-functions/nl/) (Name List) a bit like `ls` -- it lists the various objects present in the namespace in which it's being run, of the type given by its right argument:

````apl
⎕NL 2 ⍝ List variables (name class 2)
````
