The _Behind_ operator (sometimes also referred to as _Reverse Compose_) was introduced in the v20 release of Dyalog APL. Like _Over_ we met earlier, it's one of those patterns that seems pointless until the penny drops, and suddenly you see applications of it everywhere. Conor Hoekstra, \@code_report, dubbed it (somewhat tongue-in-cheek, perhaps) the [most significant addition](https://www.youtube.com/watch?v=recwy_MVgho) to Dyalog APL since the introduction of trains in 2014.

_Behind_ is easy enough to model in APL yourself:

```apl
Behind ← {⍺←⍵ ⋄ (⍺⍺ ⍺)⍵⍵ ⍵} ⍝ ⍛
```
