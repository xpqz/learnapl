# The Behind operator: `⍛`

It's not every release we get a new glyph, but v20 gives us the delightful `⍛`, code-name jot-underbar, which is the dyadic Behind operator, a functional composition. In other dialects, this is sometimes also referred to as "reverse-compose". Just like with Over (`⍥`), it initially seems pointless, but once the penny drops, it's a pattern that crops up everywhere!
