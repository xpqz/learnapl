# Preface to the 2nd Edition

> The tools we use have a profound and devious influence on our thinking habits, and therefore on our thinking abilities.
\--_Edsger W. Dijkstra_

When I first sat down to write "Learn APL", Dyalog was at version 17. During its creation, I ended up using an early version of 18. A lot has happened since then -- not least me starting to work for Dyalog. A case of poacher turned game-keeper? I have toyed with the idea of updating the text several times. In retrospect, several of the chapters are weak, showing which features of APL I wasn't really using that much at the time. The chapter on namespaces is a good example of that. We're also now closing in on a version 20 of Dyalog -- and this one will bring a lot of new stuff with it, unlike version 19 which was more feature-modest.

I want to stay true to the ideas that made people appreciate the original -- pretentious quotes, opinionated subset of APL, an irreverent tone, the odd joke thrown in -- but most importantly, I want this to remain the go-to resource for the APL-curious programmer. Whilst I do now work for Dyalog, I write this still very much as a fan and user, not an employee.

## What's new?

1. A reworked chapter on Namespaces

    As mentioned, the 1st edition chapter on namespaces was weak, and barely covered the basics. I've rewritten this to do a better job.
    
    
2. APLAN -- the APL Array Notation

    Array Notation is a new feature in version 20, but it's been decades in the making. This will truly change the way that APL is written.
    
    
3. `⎕VSET` and `⎕VGET`

    Another new v20 feature that makes it easier to get and set variables in namespaces -- yes, this can be used a bit like a dictionary, but it has many other day to day uses. 

4. The fancy new Behind operator, `⍛`
    
4. A refreshed Workflow chapter, now that Link is a bit more established. 
    
Other than that, a general clean-up to match version 20.

## Thank you

Finally, I'm still overwhelmed by the positive reception "Learn APL" received. Thank you for all your feedback, and it still brings me joy each time someone approaches me at a conference or something to tell me that this is why they got into APL.
